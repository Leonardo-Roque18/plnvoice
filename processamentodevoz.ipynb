```python
def classificar_comando_ex3(texto):
    negacoes = ["não", "nunca", "jamais", "nem"]

    # Verifica se existe alguma palavra de negação
    if any(negacao in texto.split() for negacao in negacoes):
        print("Ação cancelada. Entendi que você NÃO quer executar o comando.")
        return

    if "luz" in texto and "ligar" in texto:
        print("Ligando a luz")
    else:
        print("Comando não reconhecido")
```

```python
# Teste 1: Comando positivo
classificar_comando_ex3("ligar a luz")

# Teste 2: Comando com negação
classificar_comando_ex3("não ligar a luz")

# Teste 3: Outra palavra de negação
classificar_comando_ex3("nunca ligar a luz")

# Teste 4: Mais uma palavra de negação
classificar_comando_ex3("jamais ligar a luz")
```
